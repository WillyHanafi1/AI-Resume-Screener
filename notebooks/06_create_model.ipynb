# 06 - Create and Save Final Models

Notebook ini melatih model XGBoost menggunakan TF-IDF dari resume untuk memprediksi kategori pekerjaan, lalu menyimpan `tfidf.pkl`, `encoder.pkl`, dan `xgb_model.pkl` ke folder `models/`.

```python
import os
import pandas as pd
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier

os.makedirs('../models', exist_ok=True)

print("1. Loading Data...")
df = pd.read_csv('../data/processed/resumes_cleaned.csv').dropna()
X_text = df['cleaned_text']
y_label = df['label']

print("2. Label Encoding...")
encoder = LabelEncoder()
y = encoder.fit_transform(y_label)

print("3. TF-IDF...")
tfidf = TfidfVectorizer(max_features=5000)
X = tfidf.fit_transform(X_text)

print("4. Training XGBoost...")
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
xgb_model.fit(X, y)

print("5. Saving Models...")
joblib.dump(tfidf, '../models/tfidf.pkl')
joblib.dump(encoder, '../models/encoder.pkl')
joblib.dump(xgb_model, '../models/xgb_model.pkl')

print("Models saved successfully to /models directory!")
```
